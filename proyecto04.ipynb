```python
# Etapa 1: carga y análisis inicial
# Carga el archivo Excel con pandas
import pandas as pd
import plotly.express as px

# Leer el archivo de ventas
# Asegúrate de que el archivo esté en la misma carpeta que este cuaderno
# Si la ruta es diferente, cámbiala aquí.
df = pd.read_excel('datos_ejemplo.xlsx')

# Convertir la columna Fecha a tipo fecha
# Esto permite trabajar con fechas si luego se requiere más análisis
if 'Fecha' in df.columns:
    df['Fecha'] = pd.to_datetime(df['Fecha'])

# Cálculo del total de ventas y del promedio diario de ventas
# El promedio diario se calcula sobre toda la columna de ventas
ventas_total = df['Ventas'].sum()
ventas_promedio_diaria = df['Ventas'].mean()

print(f'Total de ventas: {ventas_total:,.2f}')
print(f'Promedio de ventas diarias: {ventas_promedio_diaria:,.2f}')

# Agrupamos las ventas por categoría para la gráfica
ventas_por_categoria = df.groupby('Categoria', as_index=False)['Ventas'].sum()
ventas_por_categoria = ventas_por_categoria.sort_values('Ventas', ascending=False)

# Gráfica de barras con Plotly
fig = px.bar(
    ventas_por_categoria,
    x='Categoria',
    y='Ventas',
    title='Total de Ventas por Categoría',
    labels={'Categoria': 'Categoría', 'Ventas': 'Ventas Totales'},
    color='Categoria'
)

fig.update_layout(template='plotly_white', xaxis_tickangle=-45)
fig.show()
```
