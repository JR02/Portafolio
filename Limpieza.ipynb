```python
import pandas as pd
import numpy as np
```

```python
# Load the CSV file
df = pd.read_csv('Vehiculos.csv')

print("Original shape:", df.shape)
print("\nFirst few rows:")
print(df.head())
print("\nLast few rows (to identify footer rows):")
print(df.tail())
```

```python
# Data Cleaning
print("=" * 60)
print("DATA CLEANING PROCESS")
print("=" * 60)

# 1. Remove footer rows (keep only data rows 0-24)
df_clean = df[df['año'].apply(lambda x: isinstance(x, int) or (isinstance(x, str) and x.isdigit()))].copy()
print(f"\n1. Removed footer rows")
print(f"   New shape: {df_clean.shape}")

# 2. Rename 'column2' to 'Total_Vehiculos'
df_clean = df_clean.rename(columns={'column2': 'Total_Vehiculos'})
print(f"\n2. Renamed 'column2' to 'Total_Vehiculos'")

# 3. Convert 'año' to integer
df_clean['año'] = df_clean['año'].astype(int)

# 4. Identify counts and percentage columns
count_cols = ['Total_Vehiculos', 'Automóvilesa', 'Autobuses', 'Jeep', 'Carga', 
              'Motocicletas', 'Volteo', 'Máquinas_Pesadas', 'Otrosb']
pct_cols = ['Automóvilesa2', 'Autobuses2', 'Jeep2', 'Carga2', 
            'Motocicletas2', 'Volteo2', 'Máquinas_Pesadas2', 'Otrosb2']

# Convert count columns to int
for col in count_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce').astype('Int64')

# Convert percentage columns to float
for col in pct_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

print(f"\n3. Converted data types")
print(f"   - Count columns to integer")
print(f"   - Percentage columns to float")

# 5. Check for missing values
print(f"\n4. Missing values:")
missing = df_clean.isnull().sum()
print(missing[missing > 0])

# 6. Display cleaned data summary
print(f"\n5. Cleaned dataset summary:")
print(f"   Shape: {df_clean.shape}")
print(f"   Year range: {df_clean['año'].min()} - {df_clean['año'].max()}")
print(f"\nFirst few rows of cleaned data:")
print(df_clean.head())
print(f"\nLast few rows of cleaned data:")
print(df_clean.tail())
print(f"\nData types:")
print(df_clean.dtypes)
```

```python
# Separate counts and percentages into distinct dataframes
print("\n" + "=" * 60)
print("REORGANIZING DATA")
print("=" * 60)

# Count dataframe (absolute numbers)
df_counts = df_clean[['año', 'Total_Vehiculos', 'Automóvilesa', 'Autobuses', 'Jeep', 
                       'Carga', 'Motocicletas', 'Volteo', 'Máquinas_Pesadas', 'Otrosb']].copy()

# Rename columns for percentages (remove '2' suffix and interpret as percentages)
df_pct = df_clean[['año', 'Automóvilesa2', 'Autobuses2', 'Jeep2', 'Carga2', 
                    'Motocicletas2', 'Volteo2', 'Máquinas_Pesadas2', 'Otrosb2']].copy()

df_pct.columns = ['año', 'Automóviles_%', 'Autobuses_%', 'Jeep_%', 'Carga_%', 
                  'Motocicletas_%', 'Volteo_%', 'Máquinas_Pesadas_%', 'Otros_%']

print("\n✓ Created df_counts - Vehicle counts by type")
print(f"  Shape: {df_counts.shape}")
print(f"  Columns: {list(df_counts.columns)}")

print("\n✓ Created df_pct - Percentage distribution by vehicle type")
print(f"  Shape: {df_pct.shape}")
print(f"  Columns: {list(df_pct.columns)}")

# Basic statistics
print("\n" + "=" * 60)
print("BASIC STATISTICS")
print("=" * 60)

print("\nTotal vehicles by year:")
print(df_counts[['año', 'Total_Vehiculos']])

print("\nVehicle type counts (2024):")
vehicle_2024 = df_counts[df_counts['año'] == 2024].iloc[0, 2:]
print(vehicle_2024.sort_values(ascending=False))

print(f"\nClean datasets ready for analysis!")
print(f"- df_clean: Full cleaned dataset with both counts and percentages")
print(f"- df_counts: Vehicle counts by type and year")
print(f"- df_pct: Percentage distribution by vehicle type and year")
```

```python
# 3. Eliminar columnas que no usaremos o duplicadas de porcentaje crudo
# Nos quedamos con los valores absolutos (las primeras columnas)
columnas_interes = [
    'año', 'Automóvilesa', 'Autobuses', 'Jeep', 
    'Carga', 'Motocicletas', 'Volteo', 'Máquinas_Pesadas', 'Otrosb'
]
df_clean = df_clean[columnas_interes]
df_clean
```
