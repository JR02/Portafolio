```python
import yfinance as yf
import matplotlib.pyplot as plt
import pandas as pd
```

```python
ticker = input("Ingrese el ticker de la acción: ") # solicita al usuario que ingrese el ticker de la acción
data = yf.download([ticker], period="6mo")# descarga los datos de AAPL desde Yahoo Finance para los últimos 6 meses
```

```python
data
```

```python
Cierre = data.Close[ticker]
Cierre.plot(title=f"Precio de cierre de {ticker} en el último año")
```

## Analisis solicitado
- Contizacion maxima
- Cotizacion minima
- cotizacion media

```python
maximo = round(Cierre.max(),2)
minimo = round(Cierre.min(),2)
promedio = round(Cierre.mean(),2)
print(maximo)
print(minimo)
print(promedio)
```

```python
import pyautogui
import pyperclip
import time
import webbrowser
```

```python
webbrowser.open("https://mail.google.com/mail/u/0/#inbox") # abre la página web el gmail en este caso, puedes cambiar la url por la que desees abrir

destinatario = "juniorisaiasm@gmail.com" # reemplaza con el correo electrónico del destinatario

asunto = f"Datos de {ticker}" # asunto del correo electrónico
contenido = f"""Estos son los datos para {ticker}.
cotización máxima: {maximo}
cotización mínima: {minimo}
cotización promedio: {promedio}


""" # contenido del correo electrónico

time.sleep(6) # espera 6 segundos para que la página cargue
pyautogui.pause = 3
#pyautogui.write(destinatario) # escribe el correo electrónico del destinatario en el campo "Para"

pyautogui.click(x=2498, y=271) # hace click en la posición (2498, 271) de la pantalla, puedes cambiar las coordenadas por las que desees hacer click

pyperclip.copy(destinatario) # copia el correo electrónico del destinatario al portapapeles
pyautogui.hotkey("ctrl", "v") # pega el correo electrónico del destinatario en el campo "Para" 
pyautogui.press("tab") # presiona la tecla "Tab" para pasar al siguiente campo
pyperclip.copy(asunto) # copia el asunto del correo electrónico al portapapeles
pyautogui.hotkey("ctrl", "v") # pega el asunto del correo electrónico en el campo "Asunto"
pyautogui.press("tab") # presiona la tecla "Tab" para pasar al siguiente campo
pyperclip.copy(contenido) # copia el contenido del correo electrónico al portapapeles
pyautogui.hotkey("ctrl", "v") # pega el contenido del correo electrónico en el cuerpo del mensaje
pyautogui.hotkey("ctrl", "enter") # presiona las teclas "Ctrl" + "Enter" para enviar el correo electrónico  

pyautogui.click(x=2498, y=271) # click en la posición (2498, 271) de la pantalla, puedes cambiar las coordenadas por las que desees hacer click enviar el correo electrónico
pyautogui.hotkey("ctrl", "w") # presiona las teclas "Ctrl" + "w" para cerrar la ventana del correo electrónico

```
