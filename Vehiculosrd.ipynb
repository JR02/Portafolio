```python
try:
   import pandas as pd
except ModuleNotFoundError:
   import sys, subprocess
   subprocess.check_call([sys.executable, "-m", "pip", "install", "pandas"])
   import pandas as pd

import pyodbc

# 1. Configurar la conexión
server = 'JUNIORPC\\SQLEXPRESS'      # Cambia por el nombre de tu servidor
database = 'Parquevehicular'  # Cambia por tu base de datos

conn_str = f'Driver={{ODBC Driver 18 for SQL Server}};Server={server};Database={database};Trusted_Connection=yes;TrustServerCertificate=yes;'
conexion = pyodbc.connect(conn_str)

# 2. Escribir la consulta SQL eligiendo solo las columnas deseadas
query = """
   SELECT TOP (1000) [año]
     ,[column2]
     ,[Automóvilesa]
     ,[Autobuses]
     ,[Jeep]
     ,[Carga]
     ,[Motocicletas]
     ,[Volteo]
     ,[Máquinas_Pesadas]
     ,[Otrosb]
     ,[Automóvilesa2]
     ,[Autobuses2]
     ,[Jeep2]
     ,[Carga2]
     ,[Motocicletas2]
     ,[Volteo2]
     ,[Máquinas_Pesadas2]
     ,[Otrosb2]
  FROM [Parquevehicular].[dbo].[3-10-11-parque-vehicular-año-según-tipo(3.10]
"""

# 3. Importar los datos directamente a Python
df = pd.read_sql(query, conexion)

# Cerrar la conexión base
conexion.close()

# Mostrar las primeras filas en Python
print(df.head())
import pyodbc

# 1. Configurar la conexión
server = 'JUNIORPC\\SQLEXPRESS'      # Cambia por el nombre de tu servidor
database = 'Parquevehicular'  # Cambia por tu base de datos

conn_str = f'Driver={{ODBC Driver 18 for SQL Server}};Server={server};Database={database};Trusted_Connection=yes;' "TrustServerCertificate=yes;" 
conexion = pyodbc.connect(conn_str)

# 2. Escribir la consulta SQL eligiendo solo las columnas deseadas
query = """
   SELECT TOP (1000) [año]
      ,[column2]
      ,[Automóvilesa]
      ,[Autobuses]
      ,[Jeep]
      ,[Carga]
      ,[Motocicletas]
      ,[Volteo]
      ,[Máquinas_Pesadas]
      ,[Otrosb]
      ,[Automóvilesa2]
      ,[Autobuses2]
      ,[Jeep2]
      ,[Carga2]
      ,[Motocicletas2]
      ,[Volteo2]
      ,[Máquinas_Pesadas2]
      ,[Otrosb2]
  FROM [Parquevehicular].[dbo].[3-10-11-parque-vehicular-año-según-tipo(3.10]
"""

# 3. Importar los datos directamente a Python
df = pd.read_sql(query, conexion)

# Cerrar la conexión base
conexion.close()

# Mostrar las primeras filas en Python
print(df.head())
```

```python
df = df.convert_dtypes()# Convertir los tipos de datos a los más adecuados
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['Automóvilesa'] = df['Automóvilesa'].fillna(method='ffill')  # Rellenar con el valor anterior
df['Autobuses'] = df['Autobuses'].fillna(method='ffill')  # Rellenar con el valor anterior
df['Jeep'] = df['Jeep'].fillna(method='ffill')  # Rellenar con el valor anterior
df['Carga'] = df['Carga'].fillna(method='ffill')  # Rellenar con el valor anterior
# Reemplaza los nulos con 0 y luego convierte a entero
df['Motocicletas'] = df['Motocicletas'].fillna(0).astype(int) # Rellenar con 0 y convertir a entero a los valors nulos NaN None NA

df['Volteo'] = df['Volteo'].fillna(0).astype(int)  # Rellenar con 0 y convertir a entero
df['Máquinas_Pesadas'] = df['Máquinas_Pesadas'].fillna(method='ffill')
df['Otrosb'] = df['Otrosb'].fillna(method='ffill')
df['Automóvilesa2'] = df['Automóvilesa2'].fillna(method='ffill') 
df['column2'] = df['column2'].fillna(0).astype(int)     #  Eliminar la columna 'column2' si no es relevante
```

```python
df.isnull().sum()
```

```python
df = df.astype({'Automóvilesa': 'int64', 'Autobuses': 'int64', 'Jeep': 'int64', 'Carga': 'int64', 
                'Motocicletas': 'int64', 'Volteo': 'int64', 'Máquinas_Pesadas': 'int64', 
                'Otrosb': 'int64', 'Automóvilesa2': 'int64'}) # Convertir a enteros     
```
